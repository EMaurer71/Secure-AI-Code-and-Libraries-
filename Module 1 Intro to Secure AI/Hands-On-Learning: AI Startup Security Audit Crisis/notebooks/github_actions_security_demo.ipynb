## 1. What `security_scan.yaml` Does

The Module 1 workflow performs a **basic CI security scan** using tools such as:
- Bandit (Python security)
- Semgrep (default rules)
- Safety (dependency vulnerabilities)
- pip-audit (Python package CVEs)
- SBOM generation

GitHub Actions automatically runs this workflow on:
- every push
- every pull request
- manual workflow dispatch

If HIGH severity issues are found, the workflow **fails** and GitHub **blocks merging** the pull request.

## 3. How GitHub Uses This Workflow

When you push code to GitHub, the following happens:

1. GitHub detects the workflow in `.github/workflows/`
2. It spins up a runner (Ubuntu VM)
3. Installs Python and dependencies
4. Runs Bandit, Semgrep, Safety, pip-audit, etc.
5. Uploads reports to the Actions UI
6. Marks the workflow as **success** or **failure**

If the workflow fails, GitHub will block merging the pull request.

This is called **branch protection** and is a core part of secure development workflows.

## 8. The Semgrep GitHub Action (Upgrade Path)

The Module 1 workflow uses Semgrep in a **basic mode**.

Semgrep also provides a **dedicated GitHub Action** that offers:
- PR inline annotations
- Autofix suggestions
- SARIF upload
- Policy enforcement
- Faster scans
- Better developer experience

### Example (not used in Module 1, but recommended for upgrades):

```yaml

    config: ./Semgrep_rules
```
