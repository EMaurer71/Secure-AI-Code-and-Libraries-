## 2. What This Workflow Does

This workflow performs **seven major security checks**:

### 1. **Bandit** — Python security analysis
Scans for insecure code patterns.

### 2. **Semgrep (auto config)** — Static analysis
Uses Semgrep’s built‑in rules to detect vulnerabilities.

### 3. **PyLint** — Code quality + potential bugs
Runs on `risk_model.py`.

### 4. **pip‑audit** — Dependency CVE scanning
Finds vulnerabilities in Python packages.

### 5. **Safety** — Dependency vulnerability scanning
Cross‑checks against Safety DB.

### 6. **pip‑licenses** — License compliance
Identifies licenses for all installed packages.

### 7. **CycloneDX SBOM** — Software Bill of Materials
Generates a full SBOM for the project.

All results are saved into the `reports/` directory and uploaded as GitHub Actions artifacts.

## 8. Upgrade Path: Adding Custom Semgrep Rules

You can extend this workflow by adding your own Semgrep rules:

```yaml
run: semgrep --config ./rules --json -o reports/semgrep_ci.json
```

This allows you to enforce:
- AI/ML‑specific security rules
- Notebook scanning
- Model integrity checks
- Data pipeline security rules
