# Vulnerability 5 — Dependency Confusion & Typosquatting

This notebook demonstrates **dependency confusion** and **typosquatting** in ML projects.

Attackers publish malicious packages with names similar to popular ML libraries, hoping developers:
- Make a typo
- Copy from an untrusted blog or repo
- Install the wrong package into production


## 1. Vulnerable Requirements File

We simulate a `requirements.txt` with legitimate packages and a few malicious ones.


```python
print("="*70)
print("VULNERABILITY 5: DEPENDENCY CONFUSION")
print("="*70)

requirements_vulnerable = """
numpy==1.24.3
pandas==2.0.3
scikit-learn==1.3.0
tensorflow==2.13.0
torch==2.0.1
transformers==4.30.2
# VULNERABILITY: Typosquatting packages
pytorch-nightly-cuda==2.1.0      # Malicious package!
tensorflow-gpu-custom==2.13.1    # Not official!
"""

print("\n[Vulnerable Requirements File]\n")
print(requirements_vulnerable)
```

## 2. Attack Scenario

1. Attacker creates malicious package: `pytorch-nightly-cuda`
2. Developer makes a typo or copies from an untrusted source
3. `pip install -r requirements.txt` installs the malicious package
4. Malicious code runs during installation or import


```python
print("\n[Attack Scenario]")
print("1. Attacker creates malicious package: 'pytorch-nightly-cuda'")
print("2. Developer copies or mistypes package name")
print("3. pip install -r requirements.txt installs malware")
print("4. Malicious code runs during install or import")

print("\n⚠️  VULNERABILITY: Dependency Confusion / Typosquatting")
print("Impact:")
print("  • Malicious code execution during pip install")
print("  • Supply chain attack affecting entire organization")
print("  • Can steal credentials, backdoor models, exfiltrate data")
print("  • Hard to detect after installation")
```

## 3. Common Malicious Naming Patterns

Examples:
- `pytorch-nightly` (vs official `torch`)
- `tensorflow-gpu-custom` (vs official `tensorflow` / `tensorflow-gpu`)
- `scikit-learn-extra` (vs official `scikit-learn`)
- `transformers-optimized` (vs official `transformers`)


```python
print("\nCommon Malicious Package Patterns:")
print("  • pytorch-nightly-cuda (official: torch)")
print("  • tensorflow-gpu-custom (official: tensorflow / tensorflow-gpu)")
print("  • scikit-learn-extra (official: scikit-learn)")
print("  • transformers-optimized (official: transformers)")
```

## 4. Detection & Prevention

### 🔍 Detection
- Check package names against official registries (PyPI, vendor docs)
- Use tools that scan for suspicious or low‑reputation packages
- Review dependency trees regularly

### ✅ Prevention
- Pin exact versions
- Use hash verification (`--require-hashes`)
- Use internal package mirrors / artifact repositories
- Treat requirements files as security‑critical assets
