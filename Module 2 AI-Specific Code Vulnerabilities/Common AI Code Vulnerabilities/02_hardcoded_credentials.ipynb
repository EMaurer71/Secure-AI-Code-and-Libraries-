# Vulnerability 2 — Hardcoded Credentials in ML Code

This notebook demonstrates **hardcoded cloud credentials** inside ML training scripts.

This is extremely common in experimental AI code and is a major security risk:
- Credentials end up in Git history
- Repos get shared or made public
- Attackers scan for exposed keys


## 1. Vulnerable Training Script

We simulate a typical ML script that downloads training data from S3 using hardcoded AWS keys.


```python
import os
import subprocess

print("="*70)
print("VULNERABILITY 2: HARDCODED CREDENTIALS")
print("="*70)

vulnerable_training_code = '''
import boto3

# VULNERABILITY: Hardcoded AWS credentials
AWS_ACCESS_KEY = "AKIAIOSFODNN7EXAMPLE"
AWS_SECRET_KEY = "wJalrXUtnFEMI/K7MDENG/bPxRfiCYEXAMPLEKEY"

def load_training_data():
    s3 = boto3.client(
        's3',
        aws_access_key_id=AWS_ACCESS_KEY,
        aws_secret_access_key=AWS_SECRET_KEY
    )
    s3.download_file('my-bucket', 'training_data.csv', 'local_data.csv')
'''

print("\n[Vulnerable Code Pattern]\n")
print(vulnerable_training_code)
```

## 2. Why This Is Dangerous

Impact:
- Credentials committed to version control (Git history)
- Exposed in public repositories (GitHub, GitLab)
- Provides unauthorized access to cloud resources
- Very common in ML projects due to quick experiments


```python
print("\n⚠️  VULNERABILITY: Hardcoded Credentials")
print("Impact:")
print("  • Credentials committed to version control (Git history)")
print("  • Exposed in public repositories (GitHub, GitLab)")
print("  • Provides unauthorized access to cloud resources")
print("  • Common in ML projects due to experimental scripts")
```

## 3. Static Analysis Detection (Bandit)

Tools like **Bandit** can detect hardcoded secrets:
- `B105` / `B106` — hardcoded password / AWS key

Below we write the vulnerable code to a file and (optionally) run Bandit.


```python
with open('vulnerable_training.py', 'w') as f:
    f.write(vulnerable_training_code)

print("\n[Running Bandit Scan]")
try:
    result = subprocess.run(
        ['bandit', 'vulnerable_training.py'],
        capture_output=True,
        text=True,
        timeout=10
    )
    print(result.stdout)
except Exception:
    print("Note: Install bandit to see detection: pip install bandit")

os.remove('vulnerable_training.py')
```

## 4. Prevention

✅ Use:
- Environment variables
- Cloud IAM roles (e.g., AWS IAM instance/profile roles)
- Secrets managers (AWS Secrets Manager, HashiCorp Vault, etc.)

Never hardcode secrets in:
- Training scripts
- Notebooks
- Config files committed to Git
